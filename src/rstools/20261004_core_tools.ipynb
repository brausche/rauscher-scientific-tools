```python
from astropy.io import fits
from glob import glob
import rstools as rs
import numpy as np
import jax.numpy as jnp
```

```python
# Here is a NIRCam dark. Including the reset frame, there are 11 samples up-the-ramp.
data = '/Users/RSC/data/JWST/NIRCam/id8758_darks_cycle4'

# Prepare to fit to 1st degree
Λ = rs.Legendre(11)

# Get data
files = glob(data+'/*.fits')
with fits.open(files[0]) as hdul:
    D = hdul[1].data[0]
```

```python
# Fit
D = rs.rowcor(D, 4)
λ = jnp.einsum('ij,jkl->ikl', Λ.F, D)
```

```python
fits.PrimaryHDU(np.asarray(λ)).writeto('/Users/rsc/Desktop/tt.fits', overwrite=True)
```
